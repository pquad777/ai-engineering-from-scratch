```python
plt.figure(figsize=(10, 5))
plt.plot(df.date, df.units_sold)
plt.title("daily sales")
plt.xlabel("date")
plt.ylabel("sales")

plt.xticks(rotation=45)
plt.xticks(range(0, len(df), 10), df.date[::10])
plt.tight_layout()
plt.show()
```

```python
weekly_sales = df.resample("W", on="date")["units_sold"].sum()
plt.plot(df.W, df.units_sold)

weekly_sales.plot(figsize=(10, 5))

plt.title("Weekly Sales")
plt.xlabel("Date")
plt.ylabel("Total Sales")

plt.show()
```

```python
df["date"] = pd.to_datetime(df["date"])
weekly_sales = df.resample("W", on="date")["units_sold"].sum()

weekly_sales.plot(figsize=(10, 5))

plt.title("Weekly Sales")
plt.xlabel("Date")
plt.ylabel("Total Sales")

plt.show()
```

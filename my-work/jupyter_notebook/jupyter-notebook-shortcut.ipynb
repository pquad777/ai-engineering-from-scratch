Jupyter Notebook shortcuts

Command mode:

| 단축키 | 기능 | 설명 |
|---|---|---|
| `Shift + Enter` | 셀 실행 | 실행 후 다음 셀로 이동 |
| `A` | 위에 셀 추가 | Command mode에서 |
| `B` | 아래에 셀 추가 | Command mode에서 |
| `DD` | 셀 삭제 | D를 두 번 빠르게 |
| `M` | Markdown으로 변경 | Command mode에서 |
| `Y` | Code로 변경 | Command mode에서 |
| `Z` | 셀 작업 되돌리기 | Command mode에서 |
| `Ctrl + Shift + H` | 단축키 목록 | JupyterLab에서는 다를 수 있음 |

Edit mode:

| 단축키 | 기능 | 설명 |
|---|---|---|
| `Tab` | 자동완성 | 변수명, 함수명, 속성 등을 자동완성 |
| `Shift + Tab` | 함수 정보 표시 | 커서가 위치한 함수의 매개변수와 설명 표시 |
| `Ctrl + /` | 주석 전환 | 선택한 코드 줄을 주석 처리하거나 해제 |
